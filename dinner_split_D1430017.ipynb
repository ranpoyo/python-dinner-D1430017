```python
# ==========================================
# 作業名稱：Python 資料型別實作作業：聚餐費用試算器
# 姓名：與那嶺藍
# 學號：D1430017
# ==========================================

# --- 測試一：使用原始資料 (4人、折價券 100 元) ---
print("【測試一】")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱（去除頭尾空白並轉成小寫）
clean_name = event_name.strip().lower()

# 2. 整理餐費並轉換型別（移除 NT$、逗號與空白，轉成浮點數/數字）
clean_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
clean_people = int(people.strip())

# 3. 分步計算費用（服務費以原始餐費計算，之後再扣除折價券）
service_fee = clean_amount * service_rate
total_amount = clean_amount + service_fee - discount
per_person_cost = total_amount / clean_people

# 4. 判斷是否超過預算（布林比較）
is_over_budget = per_person_cost > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{clean_name}")
print(f"原始餐費：{clean_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{clean_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
print("-" * 30)


# --- 測試二：只將人數改成 "5"，折價券維持 100 元 ---
print("【測試二】")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "5"  # 修改人數為 "5"
service_rate = 0.10
discount = 100
budget_per_person = 300

clean_name = event_name.strip().lower()
clean_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
clean_people = int(people.strip())

service_fee = clean_amount * service_rate
total_amount = clean_amount + service_fee - discount
per_person_cost = total_amount / clean_people
is_over_budget = per_person_cost > budget_per_person

print(f"聚餐名稱：{clean_name}")
print(f"原始餐費：{clean_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{clean_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
print("-" * 30)


# --- 測試三：人數恢復為 "4"，只將折價券改成 220 元 ---
print("【測試三】")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"  # 人數恢復為 "4"
service_rate = 0.10
discount = 220  # 修改折價券為 220
budget_per_person = 300

clean_name = event_name.strip().lower()
clean_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
clean_people = int(people.strip())

service_fee = clean_amount * service_rate
total_amount = clean_amount + service_fee - discount
per_person_cost = total_amount / clean_people
is_over_budget = per_person_cost > budget_per_person

print(f"聚餐名稱：{clean_name}")
print(f"原始餐費：{clean_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{clean_people} 人")
print(f"每人費用：{per_person_cost:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
print("-" * 30)
```
